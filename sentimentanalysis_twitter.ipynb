```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Create a sample Twitter dataset for sentiment classification
data = {
    'tweet': [
        "I absolutely love this product, it is amazing and wonderful!",
        "Worst experience ever, totally disappointed and frustrated.",
        "It is just okay, nothing special or extraordinary.",
        "Brilliant service and fantastic support from the team!",
        "Terrible quality, broke on the very first day of use.",
        "Neutral feelings about this, could be better or worse.",
        "Exceeded all my expectations! Highly recommend it to everyone.",
        "Horrible customer service, never buying from them again."
    ],
    'sentiment': ['positive', 'negative', 'neutral', 'positive', 'negative', 'neutral', 'positive', 'negative']
}
df = pd.DataFrame(data)

# Vectorize text using CountVectorizer (Bag of Words model)
vectorizer = CountVectorizer(stop_words='english')
X = vectorizer.fit_transform(df['tweet'])
y = df['sentiment']
```

```python
# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train a Logistic Regression Classifier
model = LogisticRegression()
model.fit(X_train, y_train)

# Predict and evaluate accuracy
predictions = model.predict(X_test)
print(f"Model Accuracy: {accuracy_score(y_test, predictions) * 100:.2f}%\n")

# Test custom sentences
sample_tweets = [
    "I am so happy with this wonderful service!",
    "Horrible quality, I hate it."
]
sample_vectorized = vectorizer.transform(sample_tweets)
sample_preds = model.predict(sample_vectorized)

for tweet, pred in zip(sample_tweets, sample_preds):
    print(f"Tweet: '{tweet}' --> Predicted Sentiment: {pred}")
```
